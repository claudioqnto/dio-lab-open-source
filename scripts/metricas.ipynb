```python
def calcular_metricas(vp, vn, fp, fn):
    # Total de elementos N
    n = vp + vn + fp + fn
    
    # Fórmulas da Tabela 1
    sensibilidade = vp / (vp + fn) if (vp + fn) > 0 else 0
    especificidade = vn / (fp + vn) if (fp + vn) > 0 else 0
    acuracia = (vp + vn) / n if n > 0 else 0
    precisao = vp / (vp + fp) if (vp + fp) > 0 else 0
    
    # F-score utiliza Precisão (P) e Sensibilidade (S)
    p = precisao
    s = sensibilidade
    f_score = 2 * (p * s) / (p + s) if (p + s) > 0 else 0
    
    return {
        "Sensibilidade (Recall)": sensibilidade,
        "Especificidade": especificidade,
        "Acurácia": acuracia,
        "Precisão": precisao,
        "F-score (F1)": f_score
    }

# Exemplo de valores para a Matriz de Confusão
VP = 50
VN = 40
FP = 10
FN = 10

# Execução dos cálculos
metricas = calcular_metricas(VP, VN, FP, FN)

# Exibição dos resultados
print("--- Resultados das Métricas ---")
for metrica, valor in metricas.items():
    print(f"{metrica}: {valor:.4f} ({valor * 100:.2f}%)")
```
